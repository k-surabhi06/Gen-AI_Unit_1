# Unit 1 Mini Project: AI Storyteller

## Introduction
This mini project demonstrates the use of Generative AI for creative text generation. 
The system takes a starting sentence as input and generates a short story using a 
pre-trained language model from Hugging Face.


```python
!pip install transformers
```

```python
from transformers import pipeline
```

```python
story_generator = pipeline(
    "text-generation",
    model="gpt2"
)
```

```python
prompt = "Yesterday morning I was supposed to"
```

```python
story = story_generator(
    prompt,
    max_length=120,
    do_sample=True,
    temperature=0.9
)
```

```python
print("Generated Story:\n")
print(story[0]["generated_text"])
```

## Explanation and Conclusion

In this project, GPT-2 was used to generate a short creative story from a given input sentence.
GPT-2 is a decoder-only transformer model designed for text generation tasks.

The Hugging Face text-generation pipeline simplifies the implementation by handling
tokenization and inference internally. Non-deterministic sampling was enabled to allow
creative story generation.

This project successfully demonstrates the application of Generative AI concepts
covered in Unit 1.


Story Generation using DistilGPT-2


```python
# Initialize text generation pipeline using DistilGPT-2
distil_story_generator = pipeline(
    "text-generation",
    model="distilgpt2"
)
```

```python
# Generate story using DistilGPT-2
distil_story = distil_story_generator(
    prompt,
    max_length=120,
    do_sample=True,
    temperature=0.9
)
```

```python
print("Generated Story using DistilGPT-2:\n")
print(distil_story[0]["generated_text"])
```

In addition to GPT-2, DistilGPT-2 was also used to generate stories. DistilGPT-2 is a smaller and faster version of GPT-2, which produces slightly less detailed but faster outputs. This comparison highlights the trade-off between model size, speed, and output quality.
