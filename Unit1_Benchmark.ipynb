Experiment 2:Masked Language Modeling (Missing Word)

```python
from transformers import pipeline
```

BERT - Fill Mask

```python
bert_fill = pipeline("fill-mask", model="bert-base-uncased")

sentence_bert = "The goal of Generative AI is to [MASK] new content."

bert_output = bert_fill(sentence_bert)

for pred in bert_output[:5]:
    print(pred["token_str"], "→", pred["score"])
```

RoBERTa - Fill Mask

```python
roberta_fill = pipeline("fill-mask", model="roberta-base")

sentence_roberta = "The goal of Generative AI is to <mask> new content."

roberta_output = roberta_fill(sentence_roberta)

for pred in roberta_output[:5]:
    print(pred["token_str"], "→", pred["score"])
```

BART - Fill Mask

```python
bart_fill = pipeline("fill-mask", model="facebook/bart-base")

sentence_bart = "The goal of Generative AI is to <mask> new content."

bart_output = bart_fill(sentence_bart)

for pred in bart_output[:5]:
    print(pred["token_str"], "→", pred["score"])
```

Experiment 3:Question Answering

```python
from transformers import pipeline
```

```python
context = "Generative AI poses significant risks such as hallucinations, bias, and deepfakes."
question = "What are the risks?"
```

```python
bert_qa = pipeline("question-answering", model="bert-base-uncased")

bert_output = bert_qa(question=question, context=context)

print("BERT Answer:", bert_output)
```

RoBERTa - Question Answering

```python
roberta_qa = pipeline("question-answering", model="roberta-base")

roberta_output = roberta_qa(question=question, context=context)

print("RoBERTa Answer:", roberta_output)
```

BART - Question Answering

```python
bart_qa = pipeline("question-answering", model="facebook/bart-base")

bart_output = bart_qa(question=question, context=context)

print("BART Answer:", bart_output)
```
