**NLP FUNDAMENTALS**

```python
# 1. Initialize the Tokenizer
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
```

Sample sentence

```python
sample_sentence = "Transformers changed the way machines understand language."
```

Split into tokens

```python
tokens = tokenizer.tokenize(sample_sentence)
print(f"Tokens: {tokens}")
```

Convert to Token_ID

```python
token_ids = tokenizer.convert_tokens_to_ids(tokens)
print(f"Token IDs: {token_ids}")
```

**POS Tagging**

```python
nltk.download('averaged_perceptron_tagger', quiet=True)
nltk.download('punkt', quiet=True)
import nltk
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')

```

```python
pos_tags = nltk.pos_tag(nltk.word_tokenize(sample_sentence))
print(f"POS Tags: {pos_tags}")
```

**NER**

```python
# Initialize NER pipeline (It takes the token ner)
ner_pipeline = pipeline("ner", model="dbmdz/bert-large-cased-finetuned-conll03-english", aggregation_strategy="simple")
```

```python
snippet = text[:5000]
entities = ner_pipeline(snippet)

print(f"{'Entity':<20} | {'Type':<10} | {'Score':<5}")
print("-"*78)
for entity in entities:
    if entity['score'] > 0.50:
        print(f"{entity['word']:<20} | {entity['entity_group']:<10} | {entity['score']:.2f}")
```

**Question Answering**

```python
qa_pipeline = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")
```

```python
questions = [
    "What is attention in Transformers?",
    "How does pretraining help Large Language Models?",
    "Why are transformers important for Generative AI?"
]

for q in questions:
    res = qa_pipeline(question=q, context=text[:5000])
    print(f"\nQ: {q}")
    print(f"A: {res['answer']}")
```

**Masked Language Modeling**

```python
mask_filler = pipeline("fill-mask", model="bert-base-uncased")
```

```python
masked_sentence = "The goal of Generative AI is to create new [MASK]."
preds = mask_filler(masked_sentence)

for p in preds:
    print(f"{p['token_str']}: {p['score']:.2f}")
```
